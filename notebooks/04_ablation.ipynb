# Ablation Studies

Controlled experiments to validate design decisions:
1. No augmentation vs. with augmentation
2. No CLAHE vs. with CLAHE
3. Unweighted loss vs. weighted loss

```python
import numpy as np
import matplotlib.pyplot as plt
import yaml
import json
from pathlib import Path
import sys

sys.path.insert(0, str(Path('..').resolve()))

plt.rcParams['figure.dpi'] = 150
plt.rcParams['savefig.dpi'] = 150
```

## 1. Run Ablation Studies

This takes significant time. Run `python src/ablation.py` from the command line first.

```python
ablation_path = Path('../outputs/ablation_results.json')

if ablation_path.exists():
    with open(ablation_path) as f:
        ablation_results = json.load(f)
    print("Loaded ablation results")
else:
    print("No ablation results found. Run: python src/ablation.py")
    ablation_results = {}
```

## 2. Ablation Results Visualization

```python
if ablation_results:
    names = list(ablation_results.keys())
    metrics = ['iou', 'dice', 'hd95', 'boundary_f1']
    
    fig, axes = plt.subplots(1, len(metrics), figsize=(5*len(metrics), 5))
    
    for ax, metric in zip(axes, metrics):
        values = [ablation_results[n].get(metric, {}).get('mean', 0) for n in names]
        colors = plt.cm.Set2(np.linspace(0, 1, len(names)))
        bars = ax.barh(names, values, color=colors, edgecolor='black')
        
        ax.set_title(metric.upper().replace('_', ' '))
        ax.grid(True, alpha=0.3, axis='x')
        
        for bar in bars:
            width = bar.get_width()
            ax.annotate(f'{width:.3f}', xy=(width, bar.get_y() + bar.get_height() / 2),
                       xytext=(3, 0), textcoords='offset points', ha='left', va='center')
    
    plt.suptitle('Ablation Study Results', fontsize=14)
    plt.tight_layout()
    plt.savefig('../outputs/figures/ablation_comparison.png', bbox_inches='tight')
    plt.show()
else:
    print("Run ablation studies first")
```
